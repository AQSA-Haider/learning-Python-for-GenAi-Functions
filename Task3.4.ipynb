## Using map(): apply GST to list of prices-
## Given list:

```python
prices = [100,250,400,1200,50]
```

## Creating a lambda function for GST:

```python
gst =lambda price:price+(0.18*price)
```

## Creating a new list after applying GST using map():

```python
price_after_gst = list(map(gst, prices))
```

## Printing both Lists - Original prices and the price_after_GST:

```python
print("Original prices: ",prices)
print("Price after GST: ", price_after_gst)
```
