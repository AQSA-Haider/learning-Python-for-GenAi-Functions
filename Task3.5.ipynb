## Using filter(): filter expensive products

## Printing the given list

```python
prices = [100,250,400,1200,50,2000,850]
```

## creating a function of prices greater than 500:

```python
def first(value):
    if value>500:
      return True
```

## creating a function of prices smaller than 500:

```python
def second(value):
    if value<500:
      return True
```

## Printing both the lists:

```python
print(list(filter(first,prices)))
print(list(filter(second,prices)))
```
