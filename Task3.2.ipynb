## Recursive Funtion:Factorial utility

## Creating a recursive function(factorial(n)) that:
## 1. returns factorial of n
## 2. handles two edge cases n=0 and n=1.
## 3. print error massage if n is negetive.

```python
def factorial(n):
    """This function takes an input number from the user and returns the output as the factorial of that number.
    Input= must be a positive integer.
    Output=factorial of input integer."""
    if n<0:
        print("Error: the number you entered must be a positive integer")
    elif n==0 or n==1:
        print(f"the factorial of {n} is : 1")
    else:
        factorial=1
        for pro in range(1,n+1):
            factorial=pro*factorial
            print(f"the factorial of {n} is :{factorial}")     
```

## testing the funcion:
## 1.n=1

```python
factorial(1)
```

## 2. n=-3

```python
factorial(-3)
```

## 3. n=0

```python
factorial(0)
```
