# Lambda Function(GST Calculater):

# creating a lambda function that prints the price after adding gst to it:

```python
gst=lambda price:price+(0.18*price)
print(f"the price after adding gst: {gst(500)}")
```

# optional: Creating another lambda function that prints the final price after gst and discount.

```python
final_price=lambda amount,discount:amount-(amount*discount/100)+(0.18*amount)
print(f"the final price after gst and discount is : {final_price(1000,10)}")
```
