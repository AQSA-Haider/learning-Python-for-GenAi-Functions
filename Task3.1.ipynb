## Basic Function (Price After Discount)
## Writing a function apply_discount(price,discount) that:
## 1. returns price after discount.
## 2. if discount is missing- takes a default value of 5%.
## 3. the discount cannot exceed 60%.

```python

def apply_discount(price,discount=5):
    if(discount<60):
       if discount is None:
          print(f"the price after discount is {price-(price*5/100)}")
       else:
          print(f"the price after discount is {price-(price*discount/100)}")
    else:
        print("discount sould not exceed 60%")
```

## Testing the function with value (1000,10)

```python
print("Enter the price of product and the discount to be applied(optional):")
```

```python
apply_discount(1000,10)
```

## testing it without giving discount value to see if it takes default value or not 

```python
apply_discount(500)
```
