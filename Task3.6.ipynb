## Combined Utility Function
## Create a function process_prices(prices)
## Use map()+lambda() to apply 10% discount to all the prices
## Use filter() to keep only discounted prices>300 in the list filtered_prices



```python
def process_prices(prices):
    discounted_prices=list(map(lambda x:round(x * 0.9,2),prices))
    filtered_prices=list(filter(lambda x:x>300,discounted_prices))
    print("The prices after 10% discount: ",discounted_prices)
    print("discounted prices >300: ",filtered_prices)
    
```

## Return both the lists:
## #discounted_prices
## #filtered_prices
## Testing with: the list(process_prices)


```python
process_prices=[100,500,900,50,300]
process_prices(process_prices)
```
