## Mini Problem: Menu using functions
## Create following three small functions:
## add_price(price,price_list)-- adds price to the list


```python
def add_price(price_list,price):
    price_list.append(price)
    
```

## get_average_price(prices)--> returns average prices.

```python
def get_avg_price(prices_list):
    if not prices_list:
        return 0
    return sum(prices_list)/len(prices_list)
```

## get_max_price(prices)--> returns maximum prices.

```python
def get_max_prices(prices_list):
    if not prices_list:
        return 0
    return max(prices_list)
```

## Creating a small menu:
## 1 --> Adds price
## 2 --> Shows average price
## 3 --> Shows maximum price
## q --> Quits

```python
prices=[]
while True:
    print("\n---Menu---")
    print("1 --> Show average prices")
    print("3 --> Show highest prices")
    print("q -> Quit")

    choice = input("Enter choice: ").strip().lower()

    if choice == '1':
        try:
            val = int(input("Enter prices: "))
            add_price(prices,val)
            print(f"Prices {val} added successfully.")
        except ValueError:
            print("Invalid input.Please enter a numerical value.")
    elif choice == '2':
        if prices:
            avg = get_avg_price(prices)
            print(f"Average price:{avg:.2f}")
        else:
            print("No prices in list yet.")
    elif choice == '3':
        if prices:
            highest = get_max_prices(prices)
            print(f"Highest price: {highest}")
        else:
            print("No price in the list yet.")
    elif choice == 'q':
        print("Existing program.")
        break
    else:
        print("Invalid choice,please select 1, 2, 3, or q.")

    
```
